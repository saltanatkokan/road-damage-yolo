```python
import shutil #4

total, used, free = shutil.disk_usage(PROJECT_ROOT.anchor)
print(f"Диск: {PROJECT_ROOT.anchor}")
print(f"Жалпы: {total/1e9:.1f} ГБ | Бос: {free/1e9:.1f} ГБ")

if free/1e9 < 30:
    print("⚠️ Бос орын 30 ГБ-тан аз. Деректер мен оқыту үшін тарлау болуы мүмкін.")
else:
    print("✅ Орын жеткілікті (алдын ала шек: 30 ГБ).")
```

```python
import torch, time #9

print("PyTorch нұсқасы          :", torch.__version__)
print("CUDA (PyTorch құрастырған):", torch.version.cuda)
print("cuda.is_available()      :", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU саны                 :", torch.cuda.device_count())
    print("GPU атауы                :", torch.cuda.get_device_name(0))
    props = torch.cuda.get_device_properties(0)
    print(f"GPU жады                 : {props.total_memory/1e9:.1f} ГБ")
    print("Compute capability       :", f"{props.major}.{props.minor}")

    x = torch.randn(4000, 4000, device="cuda")
    torch.cuda.synchronize()
    t0 = time.time()
    y = x @ x
    torch.cuda.synchronize()
    print(f"✅ GPU-да есептеу жұмыс істейді ({time.time()-t0:.3f} с). Құрылғы: {y.device}")
else:
    print("❌ GPU анықталмады.")
```
