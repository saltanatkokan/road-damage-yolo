Ұяшық 7.1. YOLO26s, 1 эпохалық сынақ (640)

```python
# Ұяшық 7.1 — YOLO26s, imgsz=640, 1 эпохалық сынақ
import os, time, torch
from pathlib import Path
from ultralytics import YOLO

PROJECT_ROOT = Path(r"D:\road_damage_yolo")
DATA_YAML = PROJECT_ROOT / "configs" / "data.yaml"
os.chdir(PROJECT_ROOT / "models")            # yolo26s.pt осы қалтаға жүктеледі

torch.cuda.empty_cache()
torch.cuda.reset_peak_memory_stats()

model = YOLO("yolo26s.pt")
t0 = time.time()
model.train(
    data=str(DATA_YAML), epochs=1, imgsz=640, batch=8, device=0, workers=2,
    patience=10, seed=42, amp=False,
    project=str(PROJECT_ROOT / "runs"), name="trial_yolo26s_640_e1", exist_ok=True,
)
print("\n=== ҚОРЫТЫНДЫ ===")
print(f"Жалпы уақыт: {(time.time() - t0) / 60:.1f} мин")
print(f"GPU жады (allocated, ең жоғары): {torch.cuda.max_memory_allocated() / 1e9:.2f} ГБ")
print(f"GPU жады (reserved, ең жоғары):  {torch.cuda.max_memory_reserved() / 1e9:.2f} ГБ")
```

```python
# Ұяшық 7.2 — YOLO26s, batch=4, жылдам сынақ (train жиынының 10%-ы)
import os, time, torch
from pathlib import Path
from ultralytics import YOLO

PROJECT_ROOT = Path(r"D:\road_damage_yolo")
DATA_YAML = PROJECT_ROOT / "configs" / "data.yaml"
os.chdir(PROJECT_ROOT / "models")

torch.cuda.empty_cache()
torch.cuda.reset_peak_memory_stats()

model = YOLO("yolo26s.pt")
t0 = time.time()
model.train(
    data=str(DATA_YAML), epochs=1, imgsz=640, batch=4, device=0, workers=2,
    patience=10, seed=42, amp=False, fraction=0.1,
    project=str(PROJECT_ROOT / "runs"), name="trial_yolo26s_640_b4_f01", exist_ok=True,
)
print(f"\nЖалпы уақыт: {(time.time() - t0) / 60:.1f} мин")
print(f"GPU жады (reserved, ең жоғары): {torch.cuda.max_memory_reserved() / 1e9:.2f} ГБ")
```
